```python
data['Region'] = data['Region'].fillna('Unknown')
data['Shipping_Mode'].fillna(method='ffill', inplace=True)
data['Distance_km'].fillna(data['Distance_km'].mean(), inplace=True)
data['Package_Weight_kg'].fillna(data['Package_Weight_kg'].median(), inplace=True)
data['Delivery_Rating'].fillna(data['Delivery_Rating'].mode()[0], inplace=True)
```

```python
data.duplicated().sum()
```

```python
data = data.drop_duplicates()
```

```python
data.duplicated().sum()
```

```python
data['Region'].unique()
```

```python
cat_cols = [
    "Region",
    "Shipping_Mode",
    "Vehicle_Type",
    "Weather",
    "Delivery_Status"
]

for col in cat_cols:
    data[col] = data[col].astype("string").str.strip().str.title()
```

```python
data['Region'].unique()
```

```python
Q1 = data["Distance_km"].quantile(0.25)
Q3 = data["Distance_km"].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = data[
    (data["Distance_km"] < lower) | (data["Distance_km"] > upper)
]

print(outliers)
```

```python
data["Order_Date"] = pd.to_datetime(data["Order_Date"],errors="coerce"
)
```

```python
data["Order_Year"] = data["Order_Date"].dt.year
data["Order_Month"] = data["Order_Date"].dt.month
data.head()
```

```python
data["Delivery_Delay_Days"] = ( data["Delivery_Days"]- data["Expected_Delivery_Days"])
data.head()
```

```python
data["Late_Flag"] = ( data["Delivery_Days"] > data["Expected_Delivery_Days"]).astype(int)
data.head()
```

```python
columns = [
    "Distance_km",
    "Package_Weight_kg",
    "Shipping_Cost_INR",
    "Delivery_Days"
]

for col in columns:
    data[col + "_Normalized"] = ((data[col] - data[col].min()) /(data[col].max() - data[col].min()))
```

```python
print("Shape:", data.shape)

print("Missing Values:",data.isnull().sum().sum())

print("Duplicate Rows:",data.duplicated().sum())
```
